LAB ASSIGNMENT 2: Multinomial Naive Bayes

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Load Dataset
spam_df = pd.read_csv('spam.csv', encoding='latin-1')
```

```python
# Clean columns if extra un-named columns exist
spam_df = spam_df.iloc[:, :2]
spam_df.columns = ['label', 'text']

X_text = spam_df['text']
y_text = spam_df['label']
```

```python
# Train-Test Split (consistent for fair comparison)
X_train_txt, X_test_txt, y_train_txt, y_test_txt = train_test_split(
    X_text, y_text, test_size=0.2, random_state=42, stratify=y_text
)
```

Task 1 & 2: CountVectorizer with Stop Words


```python
cv = CountVectorizer(stop_words='english')
X_train_cv = cv.fit_transform(X_train_txt)
X_test_cv = cv.transform(X_test_txt)

mnb_cv = MultinomialNB()
mnb_cv.fit(X_train_cv, y_train_txt)
y_pred_cv = mnb_cv.predict(X_test_cv)
acc_cv = accuracy_score(y_test_txt, y_pred_cv)

print("--- Lab 2 (Task 1 & 2): CountVectorizer Results ---")
print(f"Accuracy: {acc_cv:.4f}")
print(classification_report(y_test_txt, y_pred_cv))
```

Task 3: TF-IDF Vectorizer with Stop Words

```python
tfidf = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf.fit_transform(X_train_txt)
X_test_tfidf = tfidf.transform(X_test_txt)

mnb_tfidf = MultinomialNB()
mnb_tfidf.fit(X_train_tfidf, y_train_txt)
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)
acc_tfidf = accuracy_score(y_test_txt, y_pred_tfidf)

print("\n--- Lab 2 (Task 3): TF-IDF Vectorizer Results ---")
print(f"Accuracy: {acc_tfidf:.4f}")
print(classification_report(y_test_txt, y_pred_tfidf))
```

Task 4: Comparison and Conclusion

```python
print("\n" + "="*50)
print("LAB 2 CONCLUSION & COMPARISON")
print("="*50)
print(f"CountVectorizer Accuracy: {acc_cv * 100:.2f}%")
print(f"TF-IDF Vectorizer Accuracy:  {acc_tfidf * 100:.2f}%")

if acc_cv > acc_tfidf:
    print("\nConclusion: CountVectorizer is optimal for this dataset because word frequency directly indicates spam intent.")
elif acc_tfidf > acc_cv:
    print("\nConclusion: TF-IDF Vectorizer is optimal for this dataset as it penalizes overly common words and emphasizes unique spam indicators.")
else:
    print("\nConclusion: Both feature extraction techniques yielded equal performance.")
```
