Step 2: Import Required Libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from tabulate import tabulate

# Load dataset
df = pd.read_csv('voice.csv')
print("Jumlah baris dan kolom:", df.shape)
df.head()
```

Step 3: Data Preprocessing (Features, Labels, & Scaling)

```python
# Memisahkan fitur (X) dan target/label (y)
X = df.drop(columns=['label'])
y = df['label']

# Melakukan standarisasi fitur agar berskala sama (mean=0, variance=1)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

Step 4: Model Training and Evaluation across Combinations

```python
# Daftar rasio split dan kernel yang akan diuji
ratios = {
    "70:30 (Train:70, Test:30)": 0.3,
    "80:20 (Train:80, Test:20)": 0.2
}

kernels = ['linear', 'poly', 'rbf']
results = []

for ratio_name, test_size_val in ratios.items():
    # Split data berdasarkan rasio
    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled, y, test_size=test_size_val, random_state=42
    )

    for k in kernels:
        # Inisialisasi dan training model SVM
        svm_model = SVC(kernel=k, random_state=42)
        svm_model.fit(X_train, y_train)

        # Prediksi dan evaluasi akurasi
        y_pred = svm_model.predict(X_test)
        acc = accuracy_score(y_test, y_pred)

        # Simpan hasil
        results.append({
            "Data Split Ratio": ratio_name,
            "Kernel": k.upper(),
            "Accuracy": round(acc * 100, 2)
        })

# Konversi hasil ke DataFrame pandas
df_results = pd.DataFrame(results)
```

Step 5: Display the Performance Table

```python
# Menampilkan tabel menggunakan tabulate
print(tabulate(df_results, headers='keys', tablefmt='fancy_grid', showindex=False))
```
